Extracted comment data from the JSONPlaceholder REST API using Python and requests, transformed nested JSON data into a structured pandas DataFrame, performed exploratory data analysis and feature engineering, and developed visualizations to identify patterns in comment volume, text length, and email domains.

```python
import requests
import pandas as pd
from pathlib import Path
import json

# ==========================================
# Configuración de la API
# ==========================================

BASE_URL = "https://jsonplaceholder.typicode.com"

CACHE_DIR = Path("data/raw")
CACHE_DIR.mkdir(parents=True, exist_ok=True)


# ==========================================
# Función para consultar la API
# ==========================================

def fetch_json(endpoint: str, cache_name: str):
    """
    Descarga JSON desde un endpoint.

    Parámetros
    ----------
    endpoint : str
        Ruta del endpoint, por ejemplo '/users'.

    cache_name : str
        Nombre del archivo JSON para guardar en caché.

    Retorna
    -------
    data : list | dict
        Respuesta JSON convertida a objetos Python.
    """

    # Descargar desde la API
    url = f"{BASE_URL}{endpoint}"

    response = requests.get(url, timeout=30)

    # Verificar que la respuesta fue exitosa
    response.raise_for_status()

    # Convertir JSON a objetos Python
    data = response.json()

    return data


# ==========================================
# Descargar comments
# ==========================================

comments_json = fetch_json("/comments", "comments.json")


# ==========================================
# Convertir JSON a DataFrame
# ==========================================

comments = pd.json_normalize(comments_json)


# ==========================================
# Mostrar primeras filas
# ==========================================

comments.head()
```

## **Analizar los dominios de email**

Puedes extraer el dominio:

```python
comments["email_domain"] = (
    comments["email"]
    .str.split("@")
    .str[1]
)

comments["email_domain"].value_counts()
```

```python
email_domains = (
    comments["email_domain"]
    .value_counts()
    .reset_index()
)

email_domains.columns = ["email_domain", "total_comments"]

email_domains.head(10)
```
